```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import sys
import scienceplots as sp

plt.style.use(['science'])

sys.path.append('../')

from preprocessing.ng_loader import df_all_cycles
```

```python
df_real = df_all_cycles('../data/agsi_depleted/', MAXrate=6 *
                           (10**5), time_unit='months', synthetic=False)
```

```python
#This figure was made with the help of Anthropic's Claude Sonnet 5. All the code has been checked and verified.

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({'font.size': 20})


cycle_lengths = df_real.groupby(['Country', 'Site', 'Cycle_ID']).agg(Phase=('Phase', 'first'), Cycle_Length=('Cycle_position', 'max')).reset_index()


phase_order = ['injection', 'withdrawal', 'idle']

colors = {'injection': '#4E79A7', 'withdrawal': '#F28E2B', 'idle': '#E15759'}

countries = sorted(cycle_lengths['Country'].unique())
n_countries = len(countries)
n_phases = len(phase_order)

box_with = 0.25
group_width = box_with * n_phases + 0.3

fig, ax = plt.subplots(figsize=(16, 10), dpi=800)

for phase_idx, phase in enumerate(phase_order):
    positions = []
    data = []
    for country_idx, country in enumerate(countries):
        vals = cycle_lengths.loc[(cycle_lengths['Country'] == country) & (cycle_lengths['Phase'] == phase), 'Cycle_Length'].dropna().values
        pos = country_idx * group_width + phase_idx * box_with
        positions.append(pos)
        data.append(vals if len(vals) > 0 else [np.nan])  # Ensure there's at least one value to plot

    bp = ax.boxplot(data,
                    positions=positions,
                    widths=box_with*0.85,
                    patch_artist=True,
                    manage_ticks=False)

    for box in bp['boxes']:
        box.set_facecolor(colors[phase])
        box.set_alpha(0.7)
    for median in bp['medians']:
        median.set_color('black')
        x = np.mean(median.get_xdata())
        y = median.get_ydata()[0]
        ax.plot(x, y, marker='o', markersize=4,
                markerfacecolor='none', markeredgecolor='black', markeredgewidth=1,
                linestyle='none', zorder=5)

tick_positions = [i * group_width +  box_with for i in range(n_countries)]
ax.set_xticks(tick_positions)
country_labels = ['Austria',
 'Bulgaria',
 'Croatia',
 'Czech Republic',
 'Germany',
 'Hungary',
 'Italy',
 'Netherlands',
 'Poland',
 'Romania',
 'Slovakia',
 'United Kingdom']

ax.set_xticklabels(country_labels, rotation=45, ha='right')

ax.set_xlabel('Country')
ax.set_ylabel('Cycle Length (months)')
ax.set_title('Cycle Length Distribution by Country and Phase')

legend_names = ['Injection', 'Withdrawal', 'Idle']

legend_handles = [plt.Rectangle((0,0),1,1, facecolor=colors[p], alpha=0.7, edgecolor='black', linewidth=1) for p in phase_order]
ax.legend(legend_handles, legend_names, title='Phase', loc='upper right')

#ax.set_title('Cycle Length Distribution by Country and Phase')
ax.set_ylim(0, 12.5)
plt.tight_layout()
plt.show()
fig.savefig('Figures/cycle_length_distribution_by_country_and_phase.png', dpi=300)
```
