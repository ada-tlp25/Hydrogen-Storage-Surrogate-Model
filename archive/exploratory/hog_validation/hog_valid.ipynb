```python
import pandas as pd
import numpy as np

case1_data = np.loadtxt('../hogeweg_results/eclipse_caseI.csv', delimiter=';', skiprows=1)

columns = ['time[s]', 'field rate [mol/s]', 'average reservoir pressure [bar]', 'h2 rate [mol/s]', 'co2 rate [mol/s]', 'h2 rate consume [mol/s]']

case1_df = pd.DataFrame(case1_data, columns=columns)

```

```python
fig, ax = plt.subplots()
ax.plot(df_case1['YEARS'], df_case1['FPR'], label='Valid Case 1', marker='x', markersize=4)
ax.plot((case1_df['time[s]']/ (365.25 * 24 * 60 * 60)), case1_df['average reservoir pressure [bar]'], label='Truth Case 1')

ax.plot(df['YEARS'], df['FPR'], label='Valid Case 2', marker='x', markersize=4)
ax.plot((case2_df['time[s]']/ (365.25 * 24 * 60 * 60)), case2_df['average reservoir pressure [bar]'], label='Truth Case 2')
ax.set_xlabel('Time (years)')
ax.set_ylabel('Average Reservoir Pressure (bar)')
ax.grid(alpha=0.3)
ax.legend()

fig.savefig('case_comparison.png', dpi=300, bbox_inches='tight')
```
