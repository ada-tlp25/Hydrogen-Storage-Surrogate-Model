```python
import matplotlib.pyplot as plt
import numpy as np

for i in df_all['Site'].unique():
    site = df_all[df_all['Site'] == i].sort_values('Month').reset_index(drop=True)
    colors = np.where(site['Net_Injection_Withdrawal_GWh_d'] > 0, 'green', 'red')

    fig, ax = plt.subplots(figsize=(12, 6))
    ax.bar(site['Month'].dt.to_timestamp(),
           site['Net_Injection_Withdrawal_GWh_d'],
           color=colors, alpha=1, width=20,
           edgecolor='black', linewidth=0.5)
    ax.axhline(0, color='black', linewidth=0.8)
    ax.set_xlabel('Month'); ax.set_ylabel('Net (GWh/d)')
    ax.set_title(f"Net_injection_withdrawal — {site['Site'].iloc[0]}")
    plt.show()
```
